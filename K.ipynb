```python
#Фильтрация сигнала
from scipy.signal import butter, filtfilt
```

```python
#режекторный фильтр
def bandstop_filter(data, lowcut, highcut, fs, order):
    nyquist = 0.5 * fs
    low = lowcut / nyquist
    high = highcut / nyquist
    b, a = butter(order, [low, high], btype='bandstop', analog=False)
    y = filtfilt(b, a, data)
    return y
```

```python
# Параметры фильтра
fs = 1/data.TIME_s[1]  # Частота дискретизации
order = 3  # Порядок фильтра Баттерворта
```

```python
q=bandstop_filter(-data.RHEO_1_mΩ, a-0.085, a+0.085, fs, order)

f = plt.figure()
f.set_figwidth(30)
f.set_figheight(6)

plt.plot(data.TIME_s, -data.RHEO_1_mΩ, color='royalblue', label='До фильтрации')
plt.plot(data.TIME_s, q, color='red', label='После фильтрации')
plt.legend()

plt.xlabel("t, сек")
plt.ylabel("RHEO, мОм")

plt.grid(True)
plt.xlim([left, right])
```

```python
def compute_fft_spectrum(signal_data, fs):
    """Вычисление амплитудного спектра с помощью FFT"""
    N = len(signal_data)
    
    # Вычисляем FFT
    fft_result = np.fft.fft(signal_data)
    
    # Частотная ось (положительные частоты)
    freqs = np.fft.fftfreq(N, 1/fs)
    
    # Амплитудный спектр (только положительные частоты)
    magnitude = np.abs(fft_result) / N * 2  # Масштабирование
    magnitude[0] /= 2  # Коррекция для постоянной составляющей (DC)
    
    # Если N четное, нужно также скорректировать последний элемент
    if N % 2 == 0:
        magnitude[N//2] /= 2
    
    # Берем только положительные частоты (или до частоты Найквиста)
    positive_freqs = freqs[:N//2] if N % 2 == 0 else freqs[:(N//2 + 1)]
    positive_magnitude = magnitude[:len(positive_freqs)]
    
    return positive_freqs, positive_magnitude, fft_result
```

```python
#спектр исходного сигнала
freqs, magnitude, fft_result = compute_fft_spectrum(-data.RHEO_1_mΩ, fs)

#спектр сигнала после фильтрации 
f_q, magnitude_q, fft_result_q = compute_fft_spectrum(q, fs)

#спектр сигнала после удаления сплайна
f_new, magnitude_new, fft_results_new=compute_fft_spectrum(new, fs)

f = plt.figure()
f.set_figwidth(10)

plt.plot(freqs, magnitude, '--', label='Исходный сигнал', linewidth=0.5)
plt.plot(f_q, magnitude_q, label='После фильтрации')
plt.plot(f_new, magnitude_new, label='После удаления сплайна')

plt.xlim([0, 1])
plt.ylim([0, 36])
plt.grid(True)
plt.legend()
plt.xlabel('Частота, Гц')
```
